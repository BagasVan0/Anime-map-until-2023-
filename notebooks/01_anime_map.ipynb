```python
genre_dummies = df["genre"].fillna("").str.get_dummies(sep=", ")
genre_dummies = genre_dummies.drop(columns="", errors="ignore")
genre_dummies.shape
```

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

df_map = df.copy().reset_index(drop=True)
df_map["HARAM"] = genre_dummies["Hentai"] == 1
df_map["title_display"] = np.where(df_map["HARAM"], "[18+ NSFW]", df_map["title"])

num = pd.DataFrame({
    "score": df_map["score"],
    "episodes": np.log1p(df_map["episodes"]),
})
num_scaled = pd.DataFrame(MinMaxScaler().fit_transform(num), columns=num.columns)

X = pd.concat([genre_dummies.reset_index(drop=True), num_scaled], axis=1)
X.shape
```

```python
df_map["HARAM"].sum()
```

```python
df_map["score"].describe()
```

```python
df_map["score_filled"] = df_map["score"].replace(0, np.nan)
df_map["unrated"] = df_map["score_filled"].isna()
df_map["score_filled"] = df_map["score_filled"].fillna(df_map["score_filled"].median())

num = pd.DataFrame({
    "score": df_map["score_filled"],
    "episodes": np.log1p(df_map["episodes"]),
        })
num_scaled = pd.DataFrame(MinMaxScaler().fit_transform(num), columns=num.columns)

X = pd.concat([genre_dummies.reset_index(drop=True), num_scaled], axis=1)
print(X.shape, df_map["unrated"].sum())
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)
coords = pca.fit_transform(X)

df_map["pc1"] = coords[:, 0]
df_map["pc2"] = coords[:, 1]

print(pca.explained_variance_ratio_)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(df_map["pc1"], df_map["pc2"], s=5, alpha=0.5)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Anime map (PCA)")
plt.show()
```

```python
from sklearn.cluster import KMeans

km = KMeans(n_clusters=8, random_state=42, n_init=10)
df_map["cluster"] = km.fit_predict(X)
df_map["cluster"].value_counts().sort_index()
```

```python
plt.figure(figsize=(9, 7))
plt.scatter(df_map["pc1"], df_map["pc2"], c=df_map["cluster"], cmap="tab10", s=6, alpha=0.6)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Anime map (PCA + K-Means)")
plt.show()
```

```python
profile = genre_dummies.groupby(df_map["cluster"]).mean()
for c in profile.index:
    print(c, profile.loc[c].sort_values(ascending=False).head(4).round(2).to_dict())
```

```python
cluster_names = {
    0: "Mixed / Niche",
    1: "Fantasy Adventure",
    2: "Ecchi Harem Comedy",
    3: "Sci-fi & Mecha",
    4: "Comedy",
    5: "School Comedy",
    6: "Romantic Drama",
    7: "Supernatural Action",
}

df_map["cluster_name"] = df_map["cluster"].map(cluster_names)
df_map["genre_display"] = np.where(df_map["HARAM"], "-", df_map["genre"])
df_map["cluster_name"].value_counts()
```

```python
import plotly.express as px

fig = px.scatter(
    df_map, x="pc1", y="pc2", color="cluster_name",
    hover_name="title_display",
    hover_data={"score": True, "type": True, "genre_display": True, "pc1": False, "pc2": False},
    labels={"cluster_name": "Cluster", "genre_display": "Genres", "score": "Score", "type": "Type"},
    opacity=0.7, width=1000, height=700,
    title="Anime Map (PCA + K-Means)",
)

fig.update_traces(marker=dict(size=5))
fig.show()
fig.write_html("../anime_map.html")
```

```python
from sklearn.metrics import silhouette_score

for k in [4, 6, 8, 10, 12]:
    labels = KMeans(n_clusters=k, random_state=42, n_init=10).fit_predict(X)
    print(k, round(silhouette_score(X, labels), 4))
```
